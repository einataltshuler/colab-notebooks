```python
# Input for age and resting heart rate
# Both should be integer values
intAge = int(input("How old are you? "))
intRestingHR = int(input("What is your resting heart rate? "))

# Calculation for training heart rate
floatTrainingHeartRate  = 0.7*(220 - intAge) + (0.3 * intRestingHR)

# Display the training heart rate in the output
# We'll use the print function with f-strings
print(floatTrainingHeartRate)
print(f"The target training heart rate should be {floatTrainingHeartRate: .1f} beats per minute")
```
