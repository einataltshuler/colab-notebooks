```python
# Exercise Problem #1

square_feet = float(input("Enter the total square feet of land: "))

acres = square_feet / 43560

print (f"Number of acres: {acres:,.2f}")
```

```python
#Exercise Problem #2

laptop_price = float(input("Enter the price of the laptop: $"))

sales_tax = laptop_price * 0.085
total = laptop_price + sales_tax

print(f"Sales tax: ${sales_tax:,.2f}")
print(f"Total: ${total:,.2f}")

```
